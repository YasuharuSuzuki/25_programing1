### 🎲 四択問題7 例外処理 (3点)  
- 以下のプログラムで、input()で "8" と入力すると、print()で出力される文字列はどのようになるか？

```python
try:
    i = int(input())
    print(8888 / i)
    print("okikokudai"[i])
except ZeroDivisionError as e:
    print('ゼロ除算が発生しました')
except IndexError as e:
    print('範囲外参照エラーが発生しました')
else:
    print('エラー発生なし！')
finally:
    print('例外処理を終了します')
print("このプログラムを正常終了します")
```

---
#### 📕 1.
```

```

```
例外処理を終了します
このプログラムを正常終了します
```


---
#### 📘 3.
```
1111.0
a
エラー発生なし！
例外処理を終了します
このプログラムを正常終了します
```

---
#### 📙 4.
```
例外処理を終了します
```

````python
# 正解を出力してください
print('3')
````

---

### ✅ 解説

* このプログラムでは、input() に 8 が入力された場合の出力を考えます  
  （8888 / i の結果が 1111.0 になることから分かります）。  
  処理の流れを順に見ていきます。  

  - "okikokudai" は文字数が 9（インデックス 0〜8）  
  - インデックス 8 は "a"  
  - よって、ここまで 例外は発生しません。  
- 例外が発生しなかったため、  
  - except ブロックは実行されない  
  - else ブロックが実行される  
  - finally ブロックは 必ず実行される  
  - という流れになります。  

  * 👉 よって **正解は「3」** です。  
  * 課題では、10try文/02_Exceptionクラスとfinally、else.ipnyb の 練習プログラム7 で出題されています

  - 📕 1. ゼロ除算が発生しました  
    → i は 8 なので ZeroDivisionError は発生しない  

  - 📗 2. 範囲外参照エラーが発生しました  
    → インデックス 8 は有効な範囲内のため IndexError は発生しない  

  - 📙 4. 例外処理を終了します  
    → finally だけが実行されるのは、

### 🏝️ アルゴリズム問題9 清明節の日付 (15点)

#### 📝 問題文
沖縄の伝統行事「シーミー」の元となる清明節は、二十四節気の1つです。  
清明節は通常4月4日ですが、うるう年の前年は4月5日となります。  
  
例：  
2024年（うるう年）：4月4日  
2023年（うるう年の前年）：4月5日  
2025年（通常の年）：4月4日  
  
うるう年は通常4年に一度訪れますが、100年に1度うるう年にならない年があり、400年に1度うるう年になる年があります。  
2025年 : うるう年ではない（4で割れない年はうるう年ではない）    
2024年 : うるう年（4年に1度うるう年）  
2100年 : うるう年ではない（100年に1度うるう年にならない年がある）  
2000年 : うるう年（400年に1度うるう年になる年がある）  
  
西暦(Y)が与えられた時、その年の清明節の日付を出力するプログラムを作成してください。  

---
---

#### 📘 制約

$ 1582 \leq Y \leq 9999 (Yは整数) $

---
---

#### 🔢 入力

入力は以下の形式で標準入力から与えられます

---
$ Y $

---


※input()１つで取得可能

---
---

#### 📤 出力

清明節の日付を以下の形式で出力：  
M月D日  

---
---

#### 🧾 入力例1
---
```python
2024
```
---

#### 🧪 出力例1
---
```python
4月4日
```
---
※ うるう年は4月4日になります

---
---

#### 🧾 入力例2
---
```python
2023
```
---

#### 🧪 出力例2
---
```python
4月5日
```
---
※ うるう年の前年は4月5日となります

---
---

#### 🧾 入力例3
---
```python
2099
```
---

#### 🧪 出力例3
---
```python
4月4日
```
---
※ 2100年はうるう年ではないため、2099年の清明は4月4日となります

---
---

### ここ↓にPythonコードを記述してください

````python
# ここから解答例
# 入力
Y = int(input())

# 翌年がうるう年かどうかの判定
next_year = Y + 1
is_next_leap = (next_year % 4 == 0) and \
               (next_year % 100 != 0 or next_year % 400 == 0)

# 日付の判定
if is_next_leap:
    print("4月5日")
else:
    print("4月4日")
````

### ♠️ アルゴリズム問題10 ポーカーのフラッシュ判定 (15点)

#### 📝 問題文
ポーカーの役の一つに「フラッシュ」があります。これは5枚のカードの絵柄（スート）が1種類の場合を指します。
ただし、数字は関係ありません。

与えられた5枚のカードがフラッシュかどうかを判定するプログラムを作成してください。

- J（ジャック）は11として扱います
- Q（クイーン）は12として扱います
- K（キング）は13として扱います
- A（エース）は1として扱います

---
---

#### 📘 制約
- 入力されるカードは重複しません
- 各カードの数字は 1(A) ≤ 数字 ≤ 13(K) の範囲です
  - 10,J,Q,K,A は連続した数として扱いません
- スートは ♡（ハート）、♧（クラブ）、♤（スペード）、♢（ダイヤ）のいずれかです

---
---

#### 🔢 入力
入力は以下の形式で標準入力から与えられます。

---

C1 C2 C3 C4 C5

---

- スペース区切りで5枚のカードが与えられます
- 各カードはスートと数字（または絵札）の組み合わせです

---
---

#### 📤 出力
フラッシュの場合は `YES` を、そうでない場合は `NO` を出力してください。

---
---

### 🧾 入力例1
---
```
♧10 ♧8 ♧J ♧9 ♧Q
```
---

### 🧪 出力例1
---
```
YES
```
---
- 全て同じ絵柄（スート）のため、フラッシュです

---
---

### 🧾 入力例2
---
```
♡3 ♧8 ♤J ♢9 ♧Q
```
---

### 🧪 出力例2
---
```
NO
```
---
- 異なる絵柄（スート）が混在しているため、フラッシュではありません

---
---
### ここ↓にPythonコードを記述してください

````python
# 入力を受け取る
cards = input().split()

# 絵柄（スート）部分をsetで取り出す
suits = set([card[:1] for card in cards])

# フラッシュの判定
if len(suits) == 1:
    print("YES")
else:
    print("NO")
````
