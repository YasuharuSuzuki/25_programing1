### 🧩 アルゴリズム問題11　バス運賃の計算　(9点)

#### 📝 問題文
あなたは、地域のバス会社の運賃計算プログラムを作成しています。
バスの運賃は乗車距離に応じて以下のように決められています：

- 2km以下: 100円
- 2kmを超え4km以下: 150円
- 4kmを超え8km以下: 200円
- 8kmを超える: 300円

乗車距離（km）が整数で与えられたとき、運賃を計算するプログラムを作成してください。

#### 📘 制約

- 1 ≤ N ≤ 20
- Nは整数

#### 🔢 入力
入力は以下の形式で標準入力から与えられます

---
$ N $

---

- Nは乗車距離を表す整数（km）

※input()１つで取得可能

#### 📤 出力
運賃を整数で出力してください。

---

---

#### 🧪 出力例1
---

```python
100
```
---
※ 2kmの場合の運賃は100円です

---
---

#### 🧾 入力例2
---
```python

```
---

#### 🧪 出力例2
---
```python

```
---

---


---
### ここ↓にPythonコードを記述してください

````python
# バス運賃計算プログラム

# 乗車距離を入力
distance = int(input())

# 距離に応じて運賃を計算
if distance <= 2:
    fare = 100
elif distance <= 4:
    fare = 150
elif distance <= 8:
    fare = 200
else:
    fare = 300

# 結果を出力
print(fare)
````

### 📐　アルゴリズム問題12　数当てゲームの結果判定 (12点)

#### 📝 問題文
数当てゲームで、プレイヤーは1から100までの整数の中から「正解の数字」を当てようとしています。  
プレイヤーが答えた数字が「正解の数字」よりも小さい場合は"Small"、大きい場合は"Big"というヒントが表示されます。  

「正解の数字」と、プレイヤーが答えた数の履歴が与えられます。  
プレイヤーの各回答に対して、"Small"、"Big"、"Correct"のいずれかを出力するプログラムを作成してください。  

#### 📘 制約

- 1 ≤ N ≤ 10（回答回数）
- 1 ≤ Answer ≤ 100（正解の数字）
- 1 ≤ 各回答 ≤ 100（プレイヤーの回答）

#### 🔢 入力

入力は以下の形式で標準入力から与えられます。

---

$ Answer \quad N $   
$ A_1 \quad A_2  \quad ...  \quad A_N $   

---
- $ Answer $ は正解の数字
- $ N $ は回答回数
- $ A_1 \quad A_2  \quad ...  \quad A_N $ はプレイヤーの各回答

#### 📤 出力
N行に渡って、各回答に対する判定を出力してください。  
各行には"Small"、"Big"、"Correct"のいずれかを出力します。  

---
---

#### 🧾 入力例1
---
```
50 4
25 75 38 50
```
---

#### 🧪 出力例1
---
```
Small
Big
Small
Correct
```
---

---
---

#### 🧾 入力例2
---
```
30 3
10 20 40
```
---

#### 🧪 出力例2
---
```
Small
Small
Big
```
---

---
---

### 入力例3
---
```
70 5
70 70 70 70 70
```
---

### 出力例3
---
```
Correct
Correct
Correct
Correct
Correct
```
---

---
---

### ここ↓にPythonコードを記述してください

````python
# 数当てゲーム判定プログラム

# 1行目の入力（正解の数字と回答回数）
answer, N = map(int, input().split())

# 2行目の入力（プレイヤーの回答履歴）
guesses = list(map(int, input().split()))

# 各回答を判定
for guess in guesses:
    if guess < answer:
        print("Small")
    elif guess > answer:
        print("Big")
    else:
        print("Correct")
````
