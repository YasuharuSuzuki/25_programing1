## 🚫　サンプルプログラム5　ワイルドカードのexcept節（bare except）
- except節から **例外名を省略する** とすべての例外をキャッチできます。
- この書き方は `ワイルドカードのexcept節` や `bare except` などと呼ばれます。
- Pythonの[Enhancement Proposals(機能拡張提案)](https://peps.python.org/pep-0760/)では、将来的に **🚫非推奨** になる予定です。
  - 通常のexcept節のように例外クラスのインスタンスを変数で取得できません
  - SystemExitやKeyboardInterruptなど、 **補足しないことが適切な例外も補足してしまう** 問題があります
- 各エラー例外処理を記述する際には、基本的には **次に説明する基底クラス「Exception」** を使用したexcept節を使用するべきです。

````python
try:
    divider5 = int(input())
    divided5 = 9999 / divider5
    print(f"divided5={divided5}")
except:
    print('例外発生！！！(bare except)')
````

## サンプルプログラム6　⚠️　基底クラス Exception を使用したexcept節
- システム終了（SystemExit, KeyboardInterruptなど）以外のすべての組み込み例外の基底クラスである **Exceptionをexcept節に指定する** ことで、殆どの例外を補足できます。
- [公式ドキュメント](https://docs.python.org/ja/3/library/exceptions.html#exception-hierarchy)に記載の通り、プログラム上で処理するべき例外はExceptionクラスを継承しています
- 例外クラスの階層構造

```
BaseException
 ├── BaseExceptionGroup
 ├── GeneratorExit
 ├── KeyboardInterrupt
 ├── SystemExit
 └── Exception
      ├── ArithmeticError
      │    └── ZeroDivisionError
      ├── ValueError
```

````python
try:
    divider6 = int(input())
    divided6 = 9999 / divider6
    print(f"divided6={divided6}")
except Exception as e:
    print('Exception 例外発生！！！', e)
````

## ⚠️　サンプルプログラム7　else節、finally節、例外処理後のプログラムの実行
- **try 〜 finally節** では、例外発生してもしなくても実行される処理を記述することが出来ます。
- **try 〜 else節** では、例外が発生しなかったときに実行される処理を記述することが出来ます。
- 例外が発生しなかった時は **else節とfinally節がともに実行される** ことを確認しましょう。
- 例外が発生した時は **else節が実行されず、finally節が実行される** ことを確認しましょう。

````python
try:
    divider7 = int(input())
    divided7 = 9999 / divider7
    print(f"divided7={divided7:.2f}")
except ZeroDivisionError as e:
    print('ゼロ除算発生！！！')
else:
    print('else節実行！！！')
finally:
    print('finally節実行！！！')
print('例外処理後のプログラムの実行')
````

  - "ゼロ除算発生！" と出力

  - "例外処理後のプログラムの実行" と出力
- 1を入力して、以下のように出力されることと、処理の流れを確認しましょう

- ”突然の死！”を入力して、以下のように出力されることと、処理の流れを確認しましょう
  - "finally節実行！！！" と出力
  - ValueError例外に関する情報も出力

---
### 📋 まとめ
- **else節** は正常終了しているときのみ実行される
- **finally節** はすべてのパターンで実行される
- **except節** は補足する例外が発生するときのみ実行される
- **正常終了** した時または **例外が補足** された時は、その後に続くプログラムがそのまま実行される

## 📝 練習プログラム7 (10点)
以下の処理を実装してください  
1. input() を使って **整数値 N** を受け取ります  
2. **8888をNで割った値を出力** します(小数点以下第2位まで)  
  - N がゼロの場合、値は出力されずに **ZeroDivisionError** が発生するため、`ゼロ除算が発生しました` と表示してください  
  - 例外が発生しなかった場合は、計算結果を `R` として  
  `divided7=R`  
  の形式で表示してください(小数点以下第2位まで)  
3. 文字列 `okikokudai` から **N 文字目** の文字を出力します  
  **N > 9** のときは **範囲外参照エラー(IndexError)** が発生するため、`範囲外参照エラーが発生しました` と表示してください  
4. エラーが発生しなかった場合  
  **try〜else節** を使用して `エラー発生なし！` と出力します  
5. ゼロ除算、範囲外参照エラー以外のエラーが発生した時  
  `突然の死！` などと入力すると **ValueError** が発生します  
  エラーは補足しないでおきます  
6. すべての場合において、`例外処理を終了します` と出力します  
  **try〜finally節** を使用しましょう  
7. 例外を補足した時または正常終了した時は `このプログラムを正常終了します` と出力します  

````python
try:
    N = int(input())
    divided7 = 8888 / N
    print(f"divided7={divided7:.2f}")
    print("okikokudai"[N])
except ZeroDivisionError as e:
    print('ゼロ除算が発生しました')
except IndexError as e:
    print('範囲外参照エラーが発生しました')
else:
    print('エラー発生なし！')
finally:
    print('例外処理を終了します')
print("このプログラムを正常終了します")
````
