## continue文
- while文による繰り返し実行中にcontinue文を実行すると、以降の処理を実行せずに次の繰り返し処理へ移行します。
- 比較的複雑なループ文で使うことが多いですかね

---

### サンプルプログラム3 continue文
- 変数 `num` を `12` で初期化し、whileループで `1` ずつカウントアップします。
- while文の中で最初に `num` を `+1` したあと、`num` が `3` の倍数の場合は `continue実行！ (num=n)` と表示して<b>continue文を実行</b>し、繰り返し処理を継続します。
- `num` が `3` の倍数以外の時は `while繰り返し中！ (num=n)` と表示します。(通常のループ処理を実行)
- while条件を満たさなくなった場合、ループを抜けて `while終了！ (num=n)` と表示します。

````python
print("while開始！")
num = 12
while num < 23:
    num += 1
    if (num%3) == 0:
        print(f"continue実行！ (num={num})")
        continue
    print(f"while繰り返し中！ (num={num})")
print(f"while終了！ (num={num})")
````

## while 〜 else文
- while文による繰り返しを終了した後で実行する処理をelseブロックにて記述することが出来ます。
- break文によってループを中断した時はelseブロックは実行されません。
- 普段ほとんど使用しませんが、ループを全部終わらせた時や、最初の条件に見合わない可能性がある時に使うことがあります。

---

### サンプルプログラム4　 while 〜 else文
- input()でbreak番号を受け取り、サンプルプログラム２のように表示するサンプルプログラムです。
  - 最初に`while開始！ (break_num=n)` と出力します。break番号も表示しているので確認しましょう。（nは実際のbreak番号）
  - 変数 `num` を `2` で初期化し、whileループで `1` ずつカウントアップしています。
  - `num` が `break_num` になるとif文の中が実行され、`break実行！ループを抜けます (num=n)` と表示したあと<b>break文を実行</b>してループを抜けます。この時はelse文は実行されません。
  - breakしない間は `while繰り返し中！ (num=n)` と表示します。
  - breakせずにループを終了した場合は `while 〜 elseブロック実行！ (num=n)` と表示します
  - while 〜 else 後、またはbreak文によりwhileループを抜けると `while終了！ (num=n)` と表示します。

````python
break_num = int(input())
print(f"while開始！ (break_num={break_num})")
num = 2
while num < 9:
    if num >= break_num:
        print(f"break実行！ループを抜けます (num={num})")
        break
    print(f"while繰り返し中！ (num={num})")
    num += 1
else:
    print(f"while 〜 elseブロック実行！ (num={num})")
print(f"while終了！ (num={num})")
````

### 説明
- input()にてbreak番号を受け取り、break_numに格納します。
- `while開始! (break_num=n)` でループを開始します
- num が break_num になるまでの間、`while繰り返し中！ (num=n)`が実行されます。
- break番号が `9` より小さい場合
  - num が break番号になるとif文の中が実行され、`break実行！ループを抜けます (num=n)`と表示したあと<b>break文を実行</b>してループを抜けます。
  - この時は<b>while 〜 elseブロックが実行されない</b>
- break番号が9以上の場合
  - この時はbreakが実行されないため<b>while 〜 elseブロックが実行</b>され、`while 〜 elseブロック実行！ (num=n)`と表示します。
- whileループを抜けたあと`while終了！ (num=n)`と表示します。
