### 問7 字下げ (3点)
- Pythonのプログラムにおいて、コードブロックの範囲を定義するために使用される空白文字による字下げのことを何といいますか？
  1. スペーシング
  2. タブ
  3. マージン
  4. インデント

````python
# 正解を出力してください
print('4')
````

---

### ✅ 解説

* Pythonでは、**字下げ（インデント）** によって「コードのまとまり（ブロック）」を表します。

  * たとえば、`if` や `while` の後に続く処理は、インデントして書くことで「この範囲がその中の処理です」と示します。
  * 👉 よって **正解は「4. インデント」** です。

* 他の選択肢

  * 1. **スペーシング**：文字や行の間隔を調整する一般的な用語ですが、Pythonの構文上の要素ではありません。
  * 2. **タブ**：インデントに使われることもありますが、「コードブロックを定義するための仕組み」そのものではありません。
  * 3. **マージン**：文章やページの余白を指す言葉で、プログラム構造には関係ありません。

Pythonでは、この「インデント」が **文法の一部** になっています。
字下げの位置がずれているとエラーになったり、意図しない動作になることがあるため注意が必要です。

### 問8 条件分岐 (3点)
次のように、点数によって表示するメッセージを変えるプログラムを作りたい。  

- 60点未満 → 「不合格」
- 80点未満 → 「合格」
- それ以外 → 「優秀」

正しい構文になっているものを選びなさい。

---
#### 1.

```python
if score < 60
    print("不合格")
elif score < 80
    print("合格")
else
    print("優秀")
```

---
#### 2.

```python
if score < 60:
    print("不合格")
elif score < 80:
    print("合格")
else:
    print("優秀")
```

---
#### 3.

```python
if score < 60:
    print("不合格")
else score < 80:
    print("合格")
elif:
    print("優秀")
```

---
#### 4.

```python
if score < 60:
    print("不合格")
elif score > 80:
    print("合格")
else:
    print("優秀")
```

````python
# 正解を出力してください
print('2')
````

---

### ✅ 解説

* Pythonで条件に応じて処理を分けるときは、**`if` → `elif` → `else`** の構文を使います。
  各条件の後ろには **コロン（:）** を付け、次の行を字下げして処理内容を書きます。

  👉 よって **正解は「2」** です。

* 他の選択肢

  * 1：各条件の後に「:」がなく、構文エラーになります。
  * 3：`else` に条件を付けることはできず、`elif:` の後ろにも条件式が必要です。構文として誤りです。
  * 4：条件が「`score > 80`」になっており、60点以上〜80点未満を「合格」とする意図と異なります。

### 問9 while文による繰り返し処理 (3点)
- 1から奇数を順に処理し、25になった時点でループを終了して、それまでの合計を1度だけ出力したい。  
- 次のうち、正しく動作するプログラムはどれか。  

---
#### 1.

```python
num = 1
total = 0

while num <= 25:
    total += num
    num += 2
print(total)
```

---
#### 2.

```python
num = 1
total = 0

while num < 25:
    total += num
    print(total)
    num += 2
```

---
#### 3.

```python
num = 1
total = 0

while num <= 25:
    total += num
print(total)
num += 2
```

---
#### 4.

```python
num = 1
total = 0

while num <= 25:
    total = num
    num += 2
print(total)
```

````python
# 正解を出力してください
print('1')
````

---

### ✅ 解説

* 奇数を順に処理するためには、`num` を **1から始めて 2ずつ増やす** 必要があります。
* また、**25以下のときだけ処理** し、**合計をループの外で1回だけ出力** することが条件です。

👉 よって **正解は「1」** です。

* 他の選択肢

  * **2**：`print(total)` がループの中にあるため、合計が毎回出力されてしまいます。
  * **3**：`num += 2` がループの外にあるため、`num` が更新されず無限ループになります。
  * **4**：`total += num` ではなく `total = num` になっており、合計ではなく最後の数値（25）だけが出力されます。
