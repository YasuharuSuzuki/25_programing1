- 四角形の面積は たてｘよこ で求まります。  
- たての長さを$H$cm、よこの値を$W$cm、求まった面積を$R$cm^2 とした場合、

```
$H$x$W$=$R$  
こたえは $R$ cm^2です  
```

と２行で表示してください。

$ 1 <= H <= 9 (Hは整数) $  
$ 1 <= W <= 9 (Wは整数) $  


#### 🔢 入力
入力は以下の形式で標準入力から与えられます

---
$ H \quad W $

---

たて、よこの値を受け取ります

#### 📤 出力

---
$H$ x $W$ = $R$  
こたえは $R$ cm^2です  

---

---
---

#### 🧪 入力例1
---

$ 2 \quad 4 $  

---

#### 🧾 出力例1
---
2 x 4 = 8  
こたえは 8 cm^2です  

---


---
---

#### 🧪 入力例2
---

$ 9 \quad 5 $  

---

#### 🧾 出力例2
