### 🎲 四択問題7 シンプルなコード (3点)  
プログラミングはシンプルなほど不具合が起きにくく、良いプログラムとされています。  
一方で、同じ条件判定を重複して行ったり、無駄に条件分岐を多く行うのは非効率とされ、あまり良くないプログラミングとされています。  

以下のプログラムと同じ動作をする、より簡潔な書き方を選択肢から選んでください  

```python
if score < 60:
    print("不合格")
elif 60 <= score < 80:
    print("合格")
elif 80 <= score:
    print("優秀")
```
  
---


#### 📕 1.
```python
if score <= 60:
    print("不合格")
elif score <= 80:
    print("合格")
else:
    print("優秀")
```
---

#### 📗 2.
```python
if score >= 80:
    print("優秀")
elif score >= 60:
    print("合格")
elif score < 60:
    print("不合格")
```
---

#### 📘 3.
```python
if score < 60:
    print("不合格")
if 60 <= score < 80:
    print("合格")
if score >= 80:
    print("優秀")
```
---

#### 📙 4.
```python
if score < 60:
    print("不合格")
elif score < 80:
    print("合格")
else:
    print("優秀")
```
---

---

### 解説
#### 元のコードの問題点
- 60 <= score < 80 という複合条件を使用
- 80 <= score という明示的な条件が必要ない場所でも使用

- 条件を順番に判定し、最後は else で処理

#### 他の選択肢の問題点
- 1) <= の使用により境界値（60点、80点）の扱いが不適切
- 2) 最後のelifで不要な条件判定（score < 60）を行っている

### 🎲 四択問題8 if文のネスト (3点)

- 以下のプログラムを実行しようとしています。

---

```python
出席日数 = int(input())
期末試験 = int(input())
課題提出 = int(input())
if 出席日数 < 10:
    print("出席日数不足によって不可となります。")
else:
    成績点数 = min(int(期末試験*0.5 + 課題提出*0.5),100)
    if 成績点数 < 60:
        print("得点不足によって不可となります。")
    else:
        print("単位取得できます。")
```

---

- 上から順に10, 39, 92と入力した場合の出力結果を選べ。   
    📕 1. 出席日数不足によって不可となります。  
    📗 2. 得点不足によって不可となります。  
    📘 3. 単位取得できます。  
    📙 4. 何も出力されない  

````python
# 正解を出力してください
print('3')
````

---

### 解説
#### 選択肢3が正しい理由
- 出席日数は `10` です。 `if 出席日数 < 10`（10 **未満** ）という条件は **偽（False）** となり、出席不足の判定はスキップされます。
- 成績は 39 × 0.5 + 92 × 0.5 = 19.5 + 46 = 65.5 となり、`int` で切り捨てられて **65点** になります。
- 65点 は `if 成績点数 < 60` の条件に当てはまらないため、`else` の処理が実行されます。

👉 よって **正解は「3」** です。

#### 1,2,4が不適切な理由
- **1)** 出席日数が **9以下** の場合に表示されます。今回は10なので「10より小さい」には当てはまりません。
- **2)** 成績が **59点以下** の場合に表示されます。今回は65点なので合格ラインです。
- **4)** このプログラムは `if-else` 構造により、条件に応じて必ずいずれかの `print` 文が実行されます。

### 🧩　アルゴリズム問題9　サイコロ判定 (15点)

#### 📝 問題文
- A君とBさんが6面サイコロを1回ずつ振りました。
- A君のサイコロの出た目を $A$、Bさんのサイコロの出た目を $B$ とします。
- 出た目が大きい方を勝者とします。
- A君が勝った場合は `A`、Bさんが勝った場合は `B`、引き分けの場合は `Draw` と出力してください。

#### 📘 制約

$ 1 \leq A, B \leq 6 $ ($A, B$ は整数)

#### 🔢 入力
入力は以下の形式で標準入力から与えられます。
2つの数値は空白で区切られています。

---
$ A \ \ B $

---


#### 📤 出力
勝者判定の結果（A, B, Draw のいずれか）を出力してください。

---
---

#### 🧾 入力例1
---
```python
5 2
```
---

#### 🧪 出力例1
---
```python
A
```
---
※ A君が5、Bさんが2のため、大きい方の A を出力します。

---
---

#### 🧾 入力例2
---
```python
1 6
```
---

#### 🧪 出力例2
---
```python
B
```
---
※ A君が1、Bさんが6のため、大きい方の B を出力します。 

---
---

#### 🧾 入力例3
---
```python
3 3
```
---

#### 🧪 出力例3
---
```python
Draw
```
---
※ 2人とも同じ値のため、 Draw を出力してください。

---
---

### ここ↓にPythonコードを記述してください

````python
# 標準入力から値を読み込み、スペースで区切って整数に変換する
A, B = map(int, input().split())

# 大小比較を行う
if A > B:
    print("A")
elif A < B:
    print("B")
else:
    print("Draw")
````

### 📐　アルゴリズム問題10 買い物割引判定B (15点)

#### 📝 問題文
あるスーパーマーケットでは、買い物の合計金額に応じて以下のような割引を行っています：

- 1000円未満：割引なし
- 1000円以上3000円未満：5%割引
- 3000円以上5000円未満：10%割引
- 5000円以上：15%割引

買い物の合計金額が与えられたときに適用される割引率を求め、割り引いた後の買い物の金額を出力するプログラムを作成してください。  
ただし、割引金額は小数点以下切り捨てとします

#### 📘 制約

$ 1 \leq A \leq 10000 (Aは整数) $

#### 🔢 入力
入力は以下の形式で標準入力から与えられます

---
$ A $

---


※input()１つで取得可能

#### 📤 出力
計算されたチケット料金を整数で出力してください。

---
---

#### 🧾 入力例1
---
```python
1000
```
---

#### 🧪 出力例1
---
```python
950
```
---

※ 1000円のときは5%割引のため、950円となります

---
---

#### 🧾 入力例2
---
```python
5111
```
---

#### 🧪 出力例2
---
```python
4345
```
---
※ 5000円を超えると15%割引となります  
※ 5111円の15%は766.65円ですが、割引金額は小数点以下切り捨てのため766円となり、買い物の金額は5111円-766円=4345円となります（4344円にならないよう注意）  

---
---

#### 🧾 入力例3
---
```python
1
```
---

#### 🧪 出力例3
---
```python
1
```
---
※ 1円のときは割引なしのため、1円となります

---
---

### ここ↓にPythonコードを記述してください

````python
# 買い物金額から割引率を判定するプログラム

# 標準入力から買い物金額を受け取る
amount = int(input())

# 金額に応じて割引率を判定
if amount < 1000:
    discount = 0
elif amount < 3000:
    discount = 0.05
elif amount < 5000:
    discount = 0.10
else:
    discount = 0.15

# 割り引いた後の買い物の金額を計算
amount -= int(amount * discount)

# 結果を出力
print(amount)
````
