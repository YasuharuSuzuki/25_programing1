### 🎲 四択問題11 break文と合計値 (3点)
- 以下のコードのうち、1から9までの奇数を順に処理し、5になった時点でループを終了して、それまでの合計を1度だけ出力するコードを選んでください
---

#### 📕 1.
```python
total = 0
for i in range(1, 10, 2):
    if i == 5:
        total += i
        break
print(total)
```

---

#### 📗 2.
```python
total = 0
for i in range(1, 10, 2):
    if i == 5:
        total += i
    break
print(total)
```

---


#### 📘 3.
```python
total = 0
for i in range(1, 10, 2):
    if i == 5:
        break
    total += i
    print(total)
```

---

#### 📙 4.
```python
total = 0
for i in range(1, 10, 2):
    if i == 5:
        break
    total += i
print(total)
```

````python
# 正解を出力してください
print('4')
````

---

### 解説
#### 選択肢4が正しい理由
- i=1,3の時はtotalに加算される
- i=5の時点でループが終了する
- ループ終了後に合計値4（1+3）が出力される

#### 1,2,3が不適切な理由
- 1) i=1,3の時にtotalに加算せず、5の時のみ加算してからbreakするため、合計が5になる
- 2) 最初のループでいきなりbreak文が実行されてしまい、合計が0のまま終了する
- 3) 合計値が途中経過（1, 4）として出力され、最終結果のみを表示できない

### 🧮　アルゴリズム問題14　偶数の合計 (15点)

#### 📝 問題文
- 整数 $N$ (開始値) と $M$ (終了値) が与えられます。
- $N$ から $M$ までの整数（ $N$ と $M$ を含む）の中で、**偶数**だけを足し合わせた合計値を出力してください。


#### 📘 制約

$ 0 \leq N \leq M \leq 1000 $ ($N, M$ は整数)

#### 🔢 入力
入力は以下の形式で標準入力から与えられます。  
2つの数値は空白で区切られています。  

---
$ N \ \ M $

---


#### 📤 出力
対象となる偶数の合計値を出力してください。   
もし範囲内に偶数がない場合は `0` を出力してください。   

---
---

#### 🧾 入力例1
---
```python
3 9
```
---

#### 🧪 出力例1
---
```python
18
```
---
※ 3から9の間にある偶数は 4, 6, 8 です。これらを足した 4+6+8=18 を出力します。

---
---

#### 🧾 入力例2
---
```python
10 14
```
---

#### 🧪 出力例2
---
```python
36
```
---
※ 範囲は開始値と終了値を含みます。対象は 10, 12, 14 となり、合計の 36 を出力します。

---
---

#### 🧾 入力例3
---
```python
1 1
```
---

#### 🧪 出力例3
---
```python
0
```
---
※ 1から1の間には偶数がないため、合計は 0 となります。

---
---

### ここ↓にPythonコードを記述してください

````python
# 標準入力から値を読み込み、スペースで区切って整数に変換する
N, M = map(int, input().split())

# 合計値を保存する変数を初期化
total = 0

# NからMまで（Mを含むため M+1 とする）繰り返す
for i in range(N, M + 1):
    # 2で割った余りが0なら偶数
    if i % 2 == 0:
        total += i

# 結果を出力
print(total)
````

### 🐳　アルゴリズム問題15 冬休みのおもち (20点)

#### 📝 問題文
- 冬休みが $D$ 日間あります（1日目から $D$ 日目まで）。
- あなたは $N$ 個のおもちを持っています。
- 1日目から食べ始め、**$K$ 日ごと**におもちを1個ずつ食べます。
    - 例：$K=2$ の場合、1日目、3日目、5日目... に食べます。
- ただし、おもちの残り個数が $0$ になったら、それ以降は食べられません。

冬休み中の各日について、おもちを食べた日は以下のように出力してください。  
`X日目: おもちを食べました！`  

冬休み（$D$ 日間）が終わった後、最終的な結果を以下のように出力してください。  
- おもちが 0 個になった場合： `おもちがなくなりました！`  
- おもちが 1 個以上残っている場合： `おもちがY個あまりました！`  

#### 📘 制約

$ 14 \leq D \leq 21 $  
$ 1 \leq N \leq 5 $  
$ 1 \leq K \leq 5 $  

入力はすべて整数

#### 🔢 入力
入力は以下の形式で標準入力から与えられます
3つの数値は空白で区切られています。

---
$ D \ \ N \ \ K $

---

#### 📤 出力
食べた日の報告と、最終結果を出力してください。

---
---

#### 🧾 入力例1
---
```python
15 3 5
```
---

#### 🧪 出力例1
---
1日目: おもちを食べました！  
6日目: おもちを食べました！  
11日目: おもちを食べました！  
おもちがなくなりました！  

---

※ 冬休みは15日間、K=5 です。  
1日目、6日目、11日目に食べます（次は16日目なので休みが終わっています）。  
ちょうど3個食べきったので「なくなりました」と表示します。  

---
---

#### 🧾 入力例2
---
```python
14 5 7
```
---

#### 🧪 出力例2
---
1日目: おもちを食べました！  
8日目: おもちを食べました！  
おもちが3個あまりました！  

---
※ 冬休みは14日間、K=7 です。  
1日目と8日目に食べます（次は15日目なので期間外）。  
持っていた5個のうち2個食べたので、3個余っています。  

---
---

### ここ↓にPythonコードを記述してください

````python
### 💡 解答例コード
# `range(開始, 終了, 増分)` を使用することで、食べる日だけを効率よくループさせることができます。

# 標準入力から値を受け取る
D, N, K = map(int, input().split())

# range(start, stop, step) を使い、
# 1日目から D+1(終了値) まで、K日飛ばしでループする
for day in range(1, D + 1, K):
    # おもちが残っているか確認
    if N > 0:
        print(f"{day}日目: おもちを食べました！")
        N -= 1
    else:
        # おもちがなくなったら、ループを抜ける（これ以上チェック不要）
        break

# 冬休み終了後の判定
if N == 0:
    print("おもちがなくなりました！")
else:
    print(f"おもちが{N}個あまりました！")
````

### 🍣　アルゴリズム問題16 回転寿司とわさび (20点)

#### 📝 問題文
- ゆるるくんは回転寿司に来ており、**5皿** の寿司を順番に取って食べます。
- 食べるネタは以下の4種類で、それぞれの価格は以下の通りです。
    - **たまご**: 100円
    - **まぐろ**: 150円
    - **サーモン**: 200円 (わさび入り)
    - **いくら**: 300円
- 寿司のネタはリスト（配列）の形式で順番に提供されます。
- ゆるるくんはわさびが苦手です。**「サーモン」を食べると、その皿の料金を加算した後に食事を終了します**（それ以降の皿は食べません）。
- サーモンが出なかった場合は、5皿すべてを食べます。

ゆるるくんが食べた寿司の **合計金額** を出力してください。

#### 📘 制約
- ネタは必ず上記の4種類（たまご, まぐろ, サーモン, いくら）のいずれかです。
- 皿の枚数は必ず5枚です

#### 🔢 入力
入力は以下の形式で標準入力から与えられます。
5つのネタがスペース区切りで並んでいます。

---
$ S_1 \ \ S_2 \ \ S_3 \ \ S_4 \ \ S_5 $

---

#### 📤 出力
合計金額を整数で出力してください。

---
---

#### 🧾 入力例1
---
```python
たまご まぐろ サーモン いくら たまご
```
---

#### 🧪 出力例1
---
450

---

たまご(100円)を食べます。(合計100円)  
まぐろ(150円)を食べます。(合計250円)  
サーモン(200円)を食べます。(合計450円) → ここで食事終了！  
以降のいくら、たまごは食べません。  

---
---

#### 🧾 入力例2
---
```python
いくら いくら いくら いくら いくら
```
---

#### 🧪 出力例2
---
1500

---

※ サーモンが出てこないため、全て食べます。  
300円 × 5皿 = 1500円  

---
---

#### 🧾 入力例3
---
```python
サーモン たまご まぐろ たまご たまご
```
---

#### 🧪 出力例3
---
200

---

※ 1皿目のサーモン(200円)を食べて、すぐに食事が終了します。
残りの4皿は食べられません。

---
---

### ここ↓にPythonコードを記述してください

````python
### 💡 解答例コード

# `for` ループの中で `if` 文を使い、サーモンだった場合に `break` でループを抜けるのがポイントです。

# 標準入力からリストとして受け取る
# 例: ['たまご', 'まぐろ', 'サーモン', 'いくら', 'たまご']
sushi_list = input().split()

total_price = 0

# リストの中身を順番にチェックする
for item in sushi_list:
    if item == "たまご":
        total_price += 100
    elif item == "まぐろ":
        total_price += 150
    elif item == "いくら":
        total_price += 300
    elif item == "サーモン":
        total_price += 200
        # サーモンを食べたらループを強制終了する
        break

print(total_price)
````
