## continue文
- while文同様にfor文による繰り返し実行中にcontinue文を実行すると、以降の処理を実行せずに繰り返し処理を継続します。
- break文と比較するとあまり使わないです。比較的複雑なループ文で使うことが多いですかね

---

### サンプルプログラム4
- 変数 `num` を `12` で初期化し、for文で `23` までループします。rangeでループの終端番号を指定できますが、その数値-1までの値でループします。
- `num` が `3` の倍数の場合は `continue実行！ (num=n)` と表示して<b>continue文を実行</b>し、繰り返し処理を継続します。
- `num` が `3` の倍数以外の時は `for繰り返し中！ (num=n)` と表示します。(通常のループ処理を実行)
- ループを終了したら、最後に `for終了！ (num=n)` と表示します。for文ではループ終端番号として指定した数値-1の数字が出力されます

````python
print("for開始！")
for num in range(12,23):
    if (num%3) == 0:
        print(f"continue実行！ (num={num})")
        continue
    print(f"for繰り返し中！ (num={num})")
print(f"for終了！ (num={num})")
````

## for 〜 else文
- while文同様にfor文による繰り返しを終了した後で実行する処理をelseブロックにて記述することが出来ます。
- break文によってループを中断した時はelseブロックは実行されません。
- 普段ほとんど使用しませんが、ループを全部終わらせた時に実行させたい処理がある時に使うことがあります。

## サンプルプログラム5
- input()でbreak番号を受け取り、サンプルプログラム3のように表示するサンプルプログラムです
  - 最初に`for開始！ (break_num=n)` と出力します。break番号も表示しているので確認しましょう。（nは実際のbreak番号）
  - forループ内で `num` が `break_num` になるとif文が実行され、`break実行！ループを抜けます (num=n)` と表示したあと<b>break文を実行</b>してループを抜けます。この時はelse文は実行されません。
  - breakしない間は `for繰り返し中！ (num=n)` と表示します。
  - breakせずにループを終了した場合は `for 〜 elseブロック実行！ (num=n)` と表示します
  - for 〜 else 後、またはbreak文によりforループを抜けると `for終了！ (num=n)` と表示します。

````python
break_num = int(input())
print(f"for開始！ (break_num={break_num})")
for num in range(2,9):
    if num >= break_num:
        print(f"break実行！ループを抜けます (num={num})")
        break
    print(f"for繰り返し中！ (num={num})")
else:
    print(f"for 〜 elseブロック実行！ (num={num})")
print(f"for終了！ (num={num})")
````

### 説明
- input()にてbreak番号を受け取り、break_numに格納します。
- `for開始! (break_num=n)` でループを開始します
- `num` が `break_num` になるまでの間、`for繰り返し中！ (num=n)` が実行されます。
- break番号が `9` より小さい場合
  - `num` がbreak番号になるとif文の中が実行され、`break実行！ループを抜けます (num=n)` と表示したあと<b>break文を実行</b>してループを抜けます。
  - この時は<b>for 〜 elseブロックが実行されない</b>
- break番号が `9` 以上の場合
  - この時はbreakが実行されないため<b>for 〜 elseブロックが実行</b>され、`for 〜 elseブロック実行！ (num=n)` と表示します。
- forループを抜けたあと `for終了！ (num=n)` と表示します。

## for文のネスト

### 多重リスト
- リストの要素としてリストを指定することもできます。リストを要素とするリストは多重リストと呼ばれます。次は二重リストの例です。

```python
lns = [['a', 'b', 'c'], ['aaa', 'bbb', 'ccc'], [10, 20, 30]]
```

- 多重リストの要素指定は複数のインデックスで行います。 前の例で外側の [] で示されるリストの2番目の要素のリスト、 すなわち ['aaa', 'bbb', 'ccc'] の最初の要素は次のように指定します。

```python
lns[1][0]
```

### サンプルプログラム6
- 2重ループを使用して、2重のリストを表示するサンプルプログラムです

````python
list1 = [['a', 'b', 'c'], ['d', 'e', 'f'], ['g', 'h', 'i'], ['j', 'k', 'l']]

for y in range(4):
    for x in range(3):
        print(f'list1の{y+1}番目の要素（リスト）の{x+1}番目の要素 = {list1[y][x]}')
````
